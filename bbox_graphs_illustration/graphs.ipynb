# Графы страниц

Узлы и связи из structure.pkl: горизонтальные — синие, вертикальные — красные. Изображения сохраняются для всех страниц книги. Если в папке уже есть PNG, сохранение останавливается.

## Настройки

```python
from pathlib import Path
import sys

PROJECT = next(
    (path for path in (Path.cwd(), *Path.cwd().parents) if (path / "src" / "visualization.py").is_file()),
    None,
)
if PROJECT is None:
    raise RuntimeError("Откройте ноутбук из папки проекта")
if str(PROJECT) not in sys.path:
    sys.path.insert(0, str(PROJECT))

from src.visualization import show_book_stage

BOOK_ID = "1953_F1562_Op11_D4479_ячмень"
BOOK = PROJECT / "data" / "processed" / f"{BOOK_ID}.zip"
OUTPUTS = PROJECT / "outputs" / "teacher_report_20261116_221119"
BOOK_OUTPUT = OUTPUTS / BOOK_ID
```

## Проверка файлов

```python
CHECKPOINT = BOOK_OUTPUT / "checkpoints" / "structure.pkl"
for path in (BOOK, CHECKPOINT):
    if not path.is_file():
        raise FileNotFoundError(path)
print("Книга:", BOOK)
print("Checkpoint:", CHECKPOINT)
```

## Сохранение изображений

```python
save_dir = BOOK_OUTPUT / "bbox_graphs_illustration" / "graphs"
if save_dir.exists() and any(save_dir.glob("*.png")):
    raise FileExistsError(f"В папке уже есть изображения: {save_dir}")
saved = show_book_stage(
    book=BOOK,
    stage="structure",
    outputs_dir=OUTPUTS,
    book_id=BOOK_ID,
    show_cell_legend=True,
    show_graph=True,
    show_boxes=False,
    show_ids=False,
    show_values=False,
    save_dir=save_dir,
    show=False,
)
print("Сохранено изображений:", len(saved))
print("Папка:", save_dir)
```
