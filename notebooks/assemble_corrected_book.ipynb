# Сборка книги из исправленных таблиц страниц

Исправьте лист «Общая таблица» в файлах страниц и сохраните их с прежними именами. Заголовки сопоставляются без учёта регистра, лишних пробелов и различий в дефисах и тире. Повторы заголовков на одной странице и формулы не поддерживаются.

```python
from pathlib import Path
import sys

PROJECT = next((p for p in (Path.cwd(), *Path.cwd().parents) if (p / "src" / "corrected_tables.py").is_file()), None)
if PROJECT is None:
    raise RuntimeError("Откройте ноутбук из проекта ussr-ocr-table")
if str(PROJECT) not in sys.path:
    sys.path.insert(0, str(PROJECT))
from src.corrected_tables import assemble_corrected_book, prepare_corrected_pages
```

## 1. Выберите папку готовой книги

Укажите фактический путь к книге внутри папки запуска.

```python
BOOK_DIR = PROJECT / "outputs" / "teacher_report_20261116_221119" / "1953_F1562_Op11_D4479_ячмень"
print(BOOK_DIR)
```

## 2. Подготовьте копии для исправления

После нового экспорта копии создаются автоматически. Для старых результатов эта ячейка создаст их один раз. Откройте папку, исправьте лист «Общая таблица» и сохраните файлы с прежними именами. Существующие копии не перезаписываются.

```python
corrected_pages = BOOK_DIR / "corrected" / "pages_xlsx"
manifest = BOOK_DIR / "corrected" / "pages_manifest.json"
if corrected_pages.is_dir() and manifest.is_file():
    print("Используются существующие копии:", corrected_pages)
elif corrected_pages.exists() or manifest.exists():
    raise RuntimeError("Набор исправлений неполный: проверьте папку и список страниц")
else:
    corrected_pages = prepare_corrected_pages(BOOK_DIR)
    print("Созданы копии:", corrected_pages)
```

## 3. Соберите книгу

Источники и конфликты сохраняются на отдельном листе. Повторная сборка создаёт новый файл.

```python
output, conflicts = assemble_corrected_book(BOOK_DIR)
print("Итоговая книга:", output)
print("JSON:", output.with_suffix(".json"))
print("Конфликтов:", conflicts)
if conflicts:
    print("Проверьте лист «Источники и конфликты» и исправленные таблицы страниц.")
```
